```python
%load_ext autoreload
%autoreload 2
from megapose.panda3d_renderer.panda3d_scene_renderer import Panda3dSceneRenderer
from megapose.panda3d_renderer.types import Panda3dLightData
from megapose.utils.conversion import convert_scene_observation_to_panda3d

from megapose.datasets.datasets_cfg import make_scene_dataset
from megapose.datasets.datasets_cfg import make_object_dataset

from megapose.visualization.bokeh_plotter import BokehPlotter
from bokeh.io import output_notebook, show; output_notebook()
from bokeh.plotting import gridplot
```

# Load a dataset
Choose between the dataset generated using shapenet or GSO. 

Each dataset has 1 million images.

MegaPose models are trained using both datasets.

```python
ds_name, obj_ds_name = 'webdataset.shapenet_1M', 'shapenet.panda3d_bam'
# ds_name, obj_ds_name = 'webdataset.gso_1M', 'gso.normalized'

object_ds = make_object_dataset(obj_ds_name)
scene_ds = make_scene_dataset(ds_name)
renderer = Panda3dSceneRenderer(object_ds)
scene_ds.load_depth = True
```

```python
from megapose.datasets.web_scene_dataset import WebSceneDataset, IterableWebSceneDataset
from pathlib import Path
scene_ds = WebSceneDataset(Path("/data/webdatasets/shapenet_1M/"))
iterator = iter(IterableWebSceneDataset(scene_ds))
```

# Load and visualize a sample

```python
# 1. Grab a sample
sample = next(iterator)
```

```python
# 2. Visualize the RGB image
plotter = BokehPlotter()
show(plotter.plot_image(sample.rgb))
```

```python
# 3. Render the scene using ground truth annotations
def render_scene(obs):
    camera_data, object_datas = convert_scene_observation_to_panda3d(
        obs.camera_data, obs.object_datas
    )
    light_datas = [
        Panda3dLightData(
            light_type='ambient',
            color=((1., 1., 1., 1)),
        ),
    ]
    renderings = renderer.render_scene(
        object_datas,
        [camera_data],
        light_datas,
        render_depth=True,
        render_normals=True,
        render_binary_mask=True,
        copy_arrays=True,
    )
    return renderings[0]

plotter = BokehPlotter()
renderings = render_scene(sample)
fig_ds_rgb = plotter.plot_image(sample.rgb)
fig_render_rgb = plotter.plot_image(renderings.rgb)
fig_normals = plotter.plot_image(renderings.normals)
fig_overlay = plotter.plot_overlay(sample.rgb, renderings.rgb)
grid = [[fig_ds_rgb, fig_render_rgb, fig_normals, fig_overlay]]
show(gridplot(grid, sizing_mode='scale_width'))
show(gridplot(grid))
```
